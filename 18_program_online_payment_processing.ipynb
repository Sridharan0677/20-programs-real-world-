# 💳 Online Payment Processing System

## 📌 1. Definition
**Advanced exception handling** uses different exception types to handle different error conditions.

## 🎯 2. Aim
To process a simulated payment and handle invalid amount, insufficient balance, invalid input, and unexpected processing errors.

## 🔄 3. Algorithm
1. Set an initial balance.
2. Ask for payment amount.
3. Validate the amount.
4. Check available balance.
5. Deduct the amount if valid.
6. Handle different exceptions.
7. Display the result.

## 💡 4. Explanation
Custom exceptions make payment errors clear. This is only a simulated payment and does not connect to any real payment service.

## 🖥️ 5. Example
Balance: ₹5000; Payment: ₹1200 → Payment successful.

## 📄 6. Sample Output
`Payment successful. Remaining balance: ₹3800.00`

## 🏁 7. Conclusion
The program demonstrates advanced exception handling in a simulated payment system.


```python
class InvalidAmountError(Exception):
    pass

class InsufficientBalanceError(Exception):
    pass

balance = 5000.0

try:
    amount = float(input("Enter payment amount: ₹"))

    if amount <= 0:
        raise InvalidAmountError("Payment amount must be greater than zero.")

    if amount > balance:
        raise InsufficientBalanceError("Insufficient balance.")

    balance -= amount
    print(f"Payment successful. Remaining balance: ₹{balance:.2f}")

except ValueError:
    print("Input Error: Please enter a valid number.")
except InvalidAmountError as e:
    print("Payment Error:", e)
except InsufficientBalanceError as e:
    print("Payment Error:", e)
except Exception as e:
    print("Unexpected Error:", e)
```
