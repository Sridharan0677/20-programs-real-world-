# ⚙️ Application Configuration System

## 📌 1. Definition
A **global variable** is defined outside functions and can be accessed by multiple functions. The `global` keyword allows a function to modify a global variable.

## 🎯 2. Aim
To use a global configuration variable such as application status or maximum login attempts, modify it inside a function, and display the updated value.

## 🔄 3. Algorithm
1. Define a global configuration variable.
2. Create a function to modify it.
3. Use the `global` keyword.
4. Change the configuration.
5. Display the updated value.

## 💡 4. Explanation
The `global` keyword tells Python that the function should modify the variable defined outside the function.

## 🖥️ 5. Example
Initial status: `Maintenance`  
Updated status: `Online`

## 📄 6. Sample Output
The changed global configuration is displayed.

## 🏁 7. Conclusion
The program demonstrates how global variables can be used for simple application configuration.


```python
application_status = "Maintenance"
max_login_attempts = 3

def update_configuration():
    global application_status, max_login_attempts
    application_status = "Online"
    max_login_attempts = 5

print("Before update:")
print("Application Status:", application_status)
print("Maximum Login Attempts:", max_login_attempts)

update_configuration()

print("\nAfter update:")
print("Application Status:", application_status)
print("Maximum Login Attempts:", max_login_attempts)
```
