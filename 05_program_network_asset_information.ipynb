# 🖥️ Network Asset Information System

## 📌 1. Definition
**Host details** are basic information about a computer or domain, such as its hostname and resolved IP address.

## 🎯 2. Aim
To accept a hostname or domain name and display its hostname, resolved IP address, and basic network information.

## 🔄 3. Algorithm
1. Import the `socket` module.
2. Accept a hostname or domain.
3. Resolve the hostname.
4. Display hostname and IP address.
5. Handle resolution errors.
6. Stop.

## 💡 4. Explanation
`socket.gethostbyname()` resolves a hostname to an IPv4 address, while `socket.getfqdn()` returns the fully qualified domain name when available.

## 🖥️ 5. Example
Input: `example.com`  
Output includes its resolved IP address.

## 📄 6. Sample Output
`Resolved IP: ...`

## 🏁 7. Conclusion
The program demonstrates basic authorized network asset information retrieval.


```python
import socket

host = input("Enter hostname or domain name: ").strip()

try:
    ip = socket.gethostbyname(host)
    fqdn = socket.getfqdn(host)
    print("Hostname:", host)
    print("Fully Qualified Name:", fqdn)
    print("Resolved IP Address:", ip)
except socket.gaierror:
    print("Error: Could not resolve the hostname.")
except OSError as e:
    print("Network Error:", e)
```
