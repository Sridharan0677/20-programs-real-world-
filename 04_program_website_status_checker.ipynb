# 🌐 Website Status Checker

## 📌 1. Definition
An **HTTP GET request** asks a web server to return a resource such as a web page.

## 🎯 2. Aim
To send an HTTP GET request to a specified website and display its status code and a short portion of the response.

## 🔄 3. Algorithm
1. Import the HTTP request library.
2. Enter a website URL.
3. Send a GET request.
4. Display the status code.
5. Display a short response portion.
6. Handle request errors.

## 💡 4. Explanation
The `requests.get()` function sends an HTTP GET request. A timeout is used so the program does not wait indefinitely.

## 🖥️ 5. Example
For a reachable site, the program may display status code `200`.

## 📄 6. Sample Output
`Status Code: 200`

## 🏁 7. Conclusion
The program demonstrates basic website status checking using an HTTP GET request.


```python
import requests

url = input("Enter website URL (example: https://example.com): ").strip()

if not url.startswith(("http://", "https://")):
    url = "https://" + url

try:
    response = requests.get(url, timeout=5)
    print("Status Code:", response.status_code)
    print("\nResponse Preview:")
    print(response.text[:300])
except requests.exceptions.RequestException as e:
    print("Request Error:", e)
```
