# 🕵️📡 Network Traffic Monitoring Lab

## 📌 1. Definition
**Network sniffing** means observing network packets. In an authorized lab, Scapy can capture permitted test traffic and inspect basic packet fields.

## 🎯 2. Aim
To capture permitted test packets in an authorized lab and display basic non-sensitive information such as protocol, source, and destination.

## 🔄 3. Algorithm
1. Install/import Scapy.
2. Capture a small number of packets on an authorized interface.
3. Check the packet protocol layers.
4. Display basic metadata.
5. Stop capture.

## 💡 4. Explanation
This example is limited to a small packet count and prints basic metadata only. Packet capture may require administrator privileges.

## 🖥️ 5. Example
The program may display protocol, source, and destination fields for captured test packets.

## 📄 6. Sample Output
`Protocol: TCP | Source: ... | Destination: ...`

## 🏁 7. Conclusion
The program demonstrates basic packet observation using Scapy in an authorized lab.


```python
# Install Scapy first if needed:
# !pip install scapy

from scapy.all import sniff, IP, TCP, UDP, ICMP

def show_packet(packet):
    if IP in packet:
        if TCP in packet:
            protocol = "TCP"
        elif UDP in packet:
            protocol = "UDP"
        elif ICMP in packet:
            protocol = "ICMP"
        else:
            protocol = "IP"
        print(f"Protocol: {protocol} | Source: {packet[IP].src} | Destination: {packet[IP].dst}")

print("Capturing 5 permitted test packets...")
sniff(count=5, prn=show_packet, store=False)
```
