# 🌐🔧 Custom Website Health Client

## 📌 1. Definition
An **HTTP client using sockets** manually creates a TCP connection, sends an HTTP request, and receives the server response.

## 🎯 2. Aim
To connect to an authorized web server using sockets, send a basic HTTP request, and display the HTTP status and headers.

## 🔄 3. Algorithm
1. Enter an authorized web host.
2. Create a TCP socket.
3. Connect to port 80.
4. Build a basic HTTP GET request.
5. Send the request.
6. Receive the response.
7. Display status and headers.
8. Close the socket.

## 💡 4. Explanation
This example is intended for a server you are authorized to test. It uses plain HTTP on port 80 for educational purposes.

## 🖥️ 5. Example
Host: `example.com`.

## 📄 6. Sample Output
The response begins with an HTTP status line such as `HTTP/1.1 200 OK`.

## 🏁 7. Conclusion
The program demonstrates how HTTP works over a TCP socket at a basic level.


```python
import socket

host = input("Enter an authorized HTTP host (example.com): ").strip()
port = 80

request = (
    f"GET / HTTP/1.1\r\n"
    f"Host: {host}\r\n"
    f"Connection: close\r\n\r\n"
)

with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as sock:
    sock.settimeout(5)
    try:
        sock.connect((host, port))
        sock.sendall(request.encode())
        response = b""
        while True:
            chunk = sock.recv(4096)
            if not chunk:
                break
            response += chunk
    except OSError as e:
        print("Connection Error:", e)
    else:
        text_response = response.decode(errors="replace")
        header_part = text_response.split("\r\n\r\n", 1)[0]
        print("HTTP Response Headers:")
        print(header_part)
```
