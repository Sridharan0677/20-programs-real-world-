# 📡 Network Availability Monitor

## 📌 1. Definition
A **ping check** tests whether a host responds to an ICMP echo request. In this lab program, availability is checked only for an authorized private network.

## 🎯 2. Aim
To check a predefined list of authorized lab IP addresses and report which devices are reachable or unavailable.

## 🔄 3. Algorithm
1. Define a small authorized list of IP addresses.
2. Send one ping request to each host.
3. Check the return code.
4. Report reachable or unavailable.
5. Stop.

## 💡 4. Explanation
The program uses the operating system's `ping` command through Python's `subprocess` module. Use only on networks you are authorized to monitor.

## 🖥️ 5. Example
`127.0.0.1` should normally be reachable on the local computer.

## 📄 6. Sample Output
`127.0.0.1 -> Reachable`

## 🏁 7. Conclusion
The program demonstrates basic availability monitoring for an authorized lab network.


```python
import platform
import subprocess

# Use only authorized/private lab addresses.
hosts = ["127.0.0.1"]

param = "-n" if platform.system().lower() == "windows" else "-c"

for host in hosts:
    try:
        result = subprocess.run(
            ["ping", param, "1", host],
            stdout=subprocess.DEVNULL,
            stderr=subprocess.DEVNULL,
            timeout=3
        )
        status = "Reachable" if result.returncode == 0 else "Unavailable"
        print(f"{host} -> {status}")
    except (subprocess.TimeoutExpired, OSError) as e:
        print(f"{host} -> Check failed: {e}")
```
