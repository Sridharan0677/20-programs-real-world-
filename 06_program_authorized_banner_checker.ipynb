# 🔎 Authorized Service Information Checker

## 📌 1. Definition
**Banner grabbing** is the process of reading service information voluntarily returned by a server, often to identify the software or protocol in use.

## 🎯 2. Aim
To connect to an authorized test service and display the banner/version information returned by that service.

## 🔄 3. Algorithm
1. Enter an authorized host and port.
2. Create a TCP socket.
3. Set a short timeout.
4. Connect to the authorized service.
5. Receive a small banner.
6. Display it.
7. Handle connection errors.
8. Close the socket.

## 💡 4. Explanation
This program is intended for **localhost or an authorized lab server only**. It reads only the information the service chooses to return.

## 🖥️ 5. Example
Host: `127.0.0.1`, Port: `8000`.

## 📄 6. Sample Output
`Service Banner: ...`

## 🏁 7. Conclusion
The program demonstrates safe service-information retrieval from an authorized test system.


```python
import socket

host = input("Enter authorized test host: ").strip()
port = int(input("Enter authorized test port: "))

sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
sock.settimeout(3)

try:
    sock.connect((host, port))
    try:
        banner = sock.recv(1024)
        print("Service Banner:", banner.decode(errors="replace").strip())
    except socket.timeout:
        print("Connected, but the service did not send a banner.")
except (ConnectionRefusedError, TimeoutError):
    print("Could not connect to the authorized service.")
except OSError as e:
    print("Socket Error:", e)
finally:
    sock.close()
```
