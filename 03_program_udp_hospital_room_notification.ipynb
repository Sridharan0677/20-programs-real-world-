# 🏥📡 Hospital Room Notification System

## 📌 1. Definition
**UDP (User Datagram Protocol)** is a connectionless transport protocol used to send messages without establishing a persistent connection.

## 🎯 2. Aim
To develop a UDP client-server application where a client sends a hospital room-status message and the server acknowledges it.

## 🔄 3. Algorithm
1. Start the server.
2. Bind the UDP socket to localhost.
3. Wait for a message.
4. Start the client.
5. Send a room-status message.
6. Server receives and acknowledges it.
7. Stop both programs.

## 💡 4. Explanation
UDP uses `SOCK_DGRAM`. This example uses `127.0.0.1`, so communication remains on the local computer.

## 🖥️ 5. Example
Client sends: `Room 105 Available`  
Server replies: `Acknowledged: Room 105 Available`

## 📄 6. Sample Output
The server displays the received room-status message.

## 🏁 7. Conclusion
The program demonstrates basic UDP client-server communication in a safe local environment.


```python
# Run this cell first to start the UDP server in a background thread.
import socket
import threading

HOST, PORT = "127.0.0.1", 5001

def server():
    sock = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
    sock.bind((HOST, PORT))
    print(f"Server listening on {HOST}:{PORT}")
    data, address = sock.recvfrom(1024)
    message = data.decode()
    print("Server received:", message)
    sock.sendto(f"Acknowledged: {message}".encode(), address)
    sock.close()

thread = threading.Thread(target=server)
thread.start()

client = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
message = "Room 105 Available"
client.sendto(message.encode(), (HOST, PORT))
response, _ = client.recvfrom(1024)
print("Client received:", response.decode())
client.close()
thread.join()
```
