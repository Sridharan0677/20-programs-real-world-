# 📁🌐 Local Document Sharing Server

## 📌 1. Definition
A **simple HTTP server** serves files from a local directory over HTTP so authorized users can access them through a browser.

## 🎯 2. Aim
To develop a simple local HTTP server that serves authorized documents from a selected local folder.

## 🔄 3. Algorithm
1. Create a folder for authorized documents.
2. Place permitted documents in it.
3. Change the working directory to that folder.
4. Start Python's built-in HTTP server.
5. Open the local server in a browser.
6. Stop the server when finished.

## 💡 4. Explanation
Python's `http.server` module provides a simple server for local testing. This example binds to `127.0.0.1` to keep access local to the same computer.

## 🖥️ 5. Example
Browser address: `http://127.0.0.1:8000`

## 📄 6. Sample Output
The browser displays files from the selected local folder.

## 🏁 7. Conclusion
The program demonstrates local document sharing using Python's built-in HTTP server.


```python
from http.server import ThreadingHTTPServer, SimpleHTTPRequestHandler
from pathlib import Path
import os
import threading

folder = Path("shared_documents")
folder.mkdir(exist_ok=True)

# Create a harmless sample document if the folder is empty.
sample = folder / "sample.txt"
if not sample.exists():
    sample.write_text("Authorized local document.\n", encoding="utf-8")

os.chdir(folder)
server = ThreadingHTTPServer(("127.0.0.1", 8000), SimpleHTTPRequestHandler)

thread = threading.Thread(target=server.serve_forever, daemon=True)
thread.start()

print("Local document server is running at http://127.0.0.1:8000")
print("Press the stop cell below when finished.")

# To stop:
# server.shutdown()
# server.server_close()
```
