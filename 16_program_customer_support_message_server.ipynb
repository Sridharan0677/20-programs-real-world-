# 💬🖥️ Customer Support Message Server

## 📌 1. Definition
A **socket server** listens for client connections, receives data, processes it, and sends a response.

## 🎯 2. Aim
To develop a local Python socket server that receives customer support messages and sends an acknowledgement.

## 🔄 3. Algorithm
1. Create a TCP socket.
2. Bind it to localhost.
3. Listen for a client.
4. Accept the connection.
5. Receive a support message.
6. Send an acknowledgement.
7. Close the connection.

## 💡 4. Explanation
TCP provides reliable communication. The example uses localhost so it remains a local demonstration.

## 🖥️ 5. Example
Client sends: `I need help with my account.`

## 📄 6. Sample Output
Server: `Message received successfully.`

## 🏁 7. Conclusion
The program demonstrates a basic customer-support message server using Python sockets.


```python
import socket
import threading

HOST, PORT = "127.0.0.1", 5002

def server():
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as s:
        s.bind((HOST, PORT))
        s.listen(1)
        print(f"Support server listening on {HOST}:{PORT}")
        conn, address = s.accept()
        with conn:
            message = conn.recv(1024).decode(errors="replace")
            print("Server received:", message)
            conn.sendall(b"Message received successfully.")

thread = threading.Thread(target=server)
thread.start()

with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as client:
    client.connect((HOST, PORT))
    client.sendall(b"I need help with my account.")
    print("Client received:", client.recv(1024).decode())

thread.join()
```
