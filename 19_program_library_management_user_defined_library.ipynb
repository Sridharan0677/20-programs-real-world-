# 📚 Library Management System

## 📌 1. Definition
A **user-defined library** is a Python module created by the programmer and imported into another program. Classes can be placed inside the module to organize reusable functionality.

## 🎯 2. Aim
To create a user-defined Python library containing `Book`, `Student`, and `Library` classes and use it for book issue, return, and availability operations.

## 🔄 3. Algorithm
1. Create a Python module named `library_system.py`.
2. Define `Book`, `Student`, and `Library` classes.
3. Add books to the library.
4. Create a student.
5. Issue an available book.
6. Return the book.
7. Display availability.

## 💡 4. Explanation
The notebook creates a module file and imports the classes from it. This demonstrates modular programming and code reuse.

## 🖥️ 5. Example
Student Ravi issues `Python Basics`, then returns it.

## 📄 6. Sample Output
`Book issued successfully.`  
`Book returned successfully.`

## 🏁 7. Conclusion
The program demonstrates how user-defined libraries and classes can be used to build a simple reusable library management system.


```python
# Create the user-defined library file
library_code = '''
class Book:
    def __init__(self, book_id, title):
        self.book_id = book_id
        self.title = title
        self.available = True

class Student:
    def __init__(self, student_id, name):
        self.student_id = student_id
        self.name = name

class Library:
    def __init__(self):
        self.books = {}

    def add_book(self, book):
        self.books[book.book_id] = book

    def issue_book(self, book_id, student):
        book = self.books.get(book_id)
        if book is None:
            return "Book not found."
        if not book.available:
            return "Book is already issued."
        book.available = False
        return f"Book '{book.title}' issued to {student.name}."

    def return_book(self, book_id):
        book = self.books.get(book_id)
        if book is None:
            return "Book not found."
        book.available = True
        return f"Book '{book.title}' returned successfully."

    def availability(self, book_id):
        book = self.books.get(book_id)
        if book is None:
            return "Book not found."
        return "Available" if book.available else "Issued"
'''

with open("library_system.py", "w", encoding="utf-8") as f:
    f.write(library_code)

from library_system import Book, Student, Library

library = Library()
library.add_book(Book(1, "Python Basics"))
library.add_book(Book(2, "Data Structures"))

student = Student("S101", "Ravi")

print(library.issue_book(1, student))
print("Availability:", library.availability(1))
print(library.return_book(1))
print("Availability:", library.availability(1))
```
