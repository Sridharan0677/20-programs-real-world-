# 🔐 Secure Login/Input System

## 📌 1. Definition
**Buffer overflow protection** includes validating input size and format before processing data. In Python, built-in string handling provides memory safety, but input validation is still important.

## 🎯 2. Aim
To validate input length and format before processing user data and demonstrate basic protection against oversized input.

## 🔄 3. Algorithm
1. Define maximum username and password lengths.
2. Accept username and password.
3. Check input lengths.
4. Validate the username format.
5. Reject invalid or oversized input.
6. Process only valid input.

## 💡 4. Explanation
The program does not perform a real buffer overflow. It demonstrates defensive validation by restricting length and allowed characters.

## 🖥️ 5. Example
Username: `student01` and password length within the allowed limit are accepted.

## 📄 6. Sample Output
`Input accepted. Login data can be processed safely.`

## 🏁 7. Conclusion
The program demonstrates basic secure input validation and the importance of checking user data before processing.


```python
import re

MAX_USERNAME_LENGTH = 20
MAX_PASSWORD_LENGTH = 32

username = input("Enter username: ")
password = input("Enter password: ")

if len(username) > MAX_USERNAME_LENGTH:
    print("Rejected: Username is too long.")
elif not re.fullmatch(r"[A-Za-z0-9_]+", username):
    print("Rejected: Username contains invalid characters.")
elif len(password) > MAX_PASSWORD_LENGTH:
    print("Rejected: Password is too long.")
elif len(password) < 8:
    print("Rejected: Password must contain at least 8 characters.")
else:
    print("Input accepted. Login data can be processed safely.")
```
