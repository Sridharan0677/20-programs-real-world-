# 🎓 Student Academic Management System

## 📌 1. Definition
A **dictionary of lists** stores related collections under meaningful keys. It is useful for organizing structured student information.

## 🎯 2. Aim
To store student names, register numbers, subjects, and marks using a dictionary containing lists and display the information in a structured format.

## 🔄 3. Algorithm
1. Start.
2. Create a dictionary with lists for student details.
3. Add student information.
4. Traverse the lists using an index.
5. Display each student's details.
6. Stop.

## 💡 4. Explanation
Each key contains a list. The same index in each list represents one student's complete record.

## 🖥️ 5. Example
Student `Ravi`, Register No. `CS101`, Subject `Python`, Mark `85`.

## 📄 6. Sample Output
A formatted table displays all student records.

## 🏁 7. Conclusion
The program demonstrates how a dictionary of lists can organize and display student academic data efficiently.


```python
students = {
    "names": ["Ravi", "Meena", "Arun"],
    "register_numbers": ["CS101", "CS102", "CS103"],
    "subjects": ["Python", "Python", "Python"],
    "marks": [85, 92, 78]
}

print("===== Student Academic Management System =====")
print(f"{'Name':<12}{'Register No.':<15}{'Subject':<12}{'Marks':<6}")
print("-" * 45)

for i in range(len(students["names"])):
    print(f"{students['names'][i]:<12}{students['register_numbers'][i]:<15}"
          f"{students['subjects'][i]:<12}{students['marks'][i]:<6}")
```
