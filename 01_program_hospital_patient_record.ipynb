# 🏥 Hospital Patient Record System

## 📌 1. Definition
**File Handling** is used to create, open, read, write, and manage files. **Exception Handling** is used to manage errors safely during program execution.

## 🎯 2. Aim
To create a patient record file, store Patient ID, Name, Age, and Diagnosis, read the records, and handle file and input exceptions.

## 🔄 3. Algorithm
1. Start the program.
2. Display the menu.
3. Get patient details.
4. Validate the input.
5. Store the record in `patient_records.txt`.
6. Read and display stored records.
7. Handle input and file exceptions.
8. Exit the program.

## 💡 4. Explanation
The program uses append mode to add patient records and read mode to display them. `try-except` blocks handle invalid input and file-related errors.

## 🖥️ 5. Example
**Input:** Patient ID `101`, Name `Ravi`, Age `45`, Diagnosis `Fever`.

**Output:** `Patient record saved successfully!`

## 📄 6. Sample File Content
```text
101,Ravi,45,Fever
102,Arun,32,Cold
```

## 🏁 7. Conclusion
The program demonstrates practical **File Handling and Exception Handling** using a simple hospital patient record system.


````python
FILE_NAME = "patient_records.txt"

def add_patient():
    try:
        patient_id = int(input("Enter Patient ID: "))
        name = input("Enter Patient Name: ").strip()
        age = int(input("Enter Patient Age: "))
        diagnosis = input("Enter Diagnosis: ").strip()

        if not name or not diagnosis:
            raise ValueError("Name and diagnosis cannot be empty.")
        if age <= 0 or age > 120:
            raise ValueError("Enter a valid age.")

        with open(FILE_NAME, "a") as file:
            file.write(f"{patient_id},{name},{age},{diagnosis}\n")
        print("Patient record saved successfully!")

    except ValueError as e:
        print("Input Error:", e)
    except PermissionError:
        print("Permission Error: Cannot access the file.")
    except OSError as e:
        print("File Error:", e)

def read_patients():
    try:
        with open(FILE_NAME, "r") as file:
            records = file.readlines()

        if not records:
            print("No patient records found.")
            return

        print("\n----- Patient Records -----")
        for record in records:
            patient_id, name, age, diagnosis = record.strip().split(",", 3)
            print(f"Patient ID : {patient_id}")
            print(f"Name       : {name}")
            print(f"Age        : {age}")
            print(f"Diagnosis  : {diagnosis}")
            print("---------------------------")

    except FileNotFoundError:
        print("Record file does not exist yet.")
    except PermissionError:
        print("Permission Error: Cannot read the file.")
    except OSError as e:
        print("File Error:", e)

while True:
    print("\n1. Add Patient Record")
    print("2. Read Patient Records")
    print("3. Exit")
    try:
        choice = int(input("Enter your choice: "))
        if choice == 1:
            add_patient()
        elif choice == 2:
            read_patients()
        elif choice == 3:
            print("Program exited.")
            break
        else:
            print("Please choose 1, 2, or 3.")
    except ValueError:
        print("Input Error: Please enter a number.")
````
