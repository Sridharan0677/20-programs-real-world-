# 🏥📄 Hospital Medical Report Reader

## 📌 1. Definition
**File exception handling** allows a program to respond safely when a requested file is missing, inaccessible, or contains invalid input.

## 🎯 2. Aim
To open and read a hospital medical-report file specified by the user and handle file-not-found, permission, and invalid input errors.

## 🔄 3. Algorithm
1. Ask for the report filename.
2. Validate that a filename was entered.
3. Open the file in read mode.
4. Display the report content.
5. Handle `FileNotFoundError`, `PermissionError`, and other errors.
6. Stop.

## 💡 4. Explanation
The `with open()` statement automatically closes the file. Exceptions provide user-friendly error messages.

## 🖥️ 5. Example
Input: `medical_report.txt`.

## 📄 6. Sample Output
The report content is displayed when the file exists.

## 🏁 7. Conclusion
The program demonstrates safe reading of a medical-report file with appropriate exception handling.


```python
filename = input("Enter medical report filename: ").strip()

try:
    if not filename:
        raise ValueError("Filename cannot be empty.")

    patient_id = input("Enter Patient ID: ")
    patient_name = input("Enter Patient Name: ")
    age = input("Enter Age: ")
    diagnosis = input("Enter Diagnosis: ")

    with open(filename, "w", encoding="utf-8") as file:
        file.write(f"Patient ID: {patient_id}\n")
        file.write(f"Patient Name: {patient_name}\n")
        file.write(f"Age: {age}\n")
        file.write(f"Diagnosis: {diagnosis}\n")

    with open(filename, "r", encoding="utf-8") as file:
        content = file.read()

    print("\n----- Medical Report -----")
    print(content)

except ValueError as e:
    print("Input Error:", e)
except PermissionError:
    print("Error: Permission denied.")
except OSError as e:
    print("File Error:", e)
```
