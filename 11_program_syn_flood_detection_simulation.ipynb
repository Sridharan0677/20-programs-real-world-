# 🛡️ SYN Flood Detection Simulation

## 📌 1. Definition
A **SYN flood** is a condition where a system receives many TCP connection-start attempts that remain incomplete. Detection systems can identify unusually high counts of incomplete connections.

## 🎯 2. Aim
To create a **safe offline simulation** that analyzes sample connection records and identifies an unusually high number of incomplete TCP connection attempts.

## 🔄 3. Algorithm
1. Create sample offline connection records.
2. Count incomplete SYN attempts.
3. Set a detection threshold.
4. Compare the count with the threshold.
5. Report normal or suspicious activity.

## 💡 4. Explanation
No packets are generated or transmitted. The program only analyzes pre-created sample data.

## 🖥️ 5. Example
If incomplete SYN attempts are greater than the threshold, the program reports a possible SYN-flood condition.

## 📄 6. Sample Output
`Alert: unusually high incomplete TCP connection count detected.`

## 🏁 7. Conclusion
The simulation demonstrates the defensive idea of detecting abnormal incomplete TCP connections without performing an attack.


```python
# Safe offline detection simulation: no network packets are generated.
connections = [
    ("10.0.0.2", "SYN_RECEIVED"),
    ("10.0.0.3", "SYN_RECEIVED"),
    ("10.0.0.4", "COMPLETED"),
    ("10.0.0.5", "SYN_RECEIVED"),
    ("10.0.0.6", "SYN_RECEIVED"),
    ("10.0.0.7", "COMPLETED"),
    ("10.0.0.8", "SYN_RECEIVED"),
]

threshold = 4
incomplete = [record for record in connections if record[1] == "SYN_RECEIVED"]

print("Total connection records:", len(connections))
print("Incomplete SYN attempts:", len(incomplete))
print("Detection threshold:", threshold)

if len(incomplete) > threshold:
    print("Alert: unusually high incomplete TCP connection count detected.")
else:
    print("Status: connection activity is within the sample threshold.")
```
