# 👥 Employee Shift Management

## 📌 1. Definition
**List slicing** extracts a selected portion of a list using the syntax `[start:stop:step]`.

## 🎯 2. Aim
To store employee shift schedules in a list and use slicing to display first, last, alternate, and specified-range shifts.

## 🔄 3. Algorithm
1. Create a list of employee shifts.
2. Display the complete list.
3. Use slicing for first shifts.
4. Use slicing for last shifts.
5. Use a step for alternate shifts.
6. Display a specified range.

## 💡 4. Explanation
Slicing allows selected elements to be accessed without changing the original list.

## 🖥️ 5. Example
`shifts[:3]` displays the first three shifts and `shifts[-3:]` displays the last three.

## 📄 6. Sample Output
Different sliced portions of the shift list are displayed.

## 🏁 7. Conclusion
The program demonstrates practical list slicing for employee shift management.


```python
shifts = [
    "Ravi - Morning", "Meena - Evening", "Arun - Night",
    "Kaviya - Morning", "Vijay - Evening", "Priya - Night",
    "Suresh - Morning", "Anu - Evening"
]

print("All Shifts:", shifts)
print("First 3 Shifts:", shifts[:3])
print("Last 3 Shifts:", shifts[-3:])
print("Alternate Shifts:", shifts[::2])
print("Shifts from index 2 to 5:", shifts[2:6])
```
